7. Sum of weights (1 point)

```python
df_asia = df[df["origin"] == "Asia"]
df_asia.head()
```

```python
df_asia_fil = df_asia[["vehicle_weight","model_year"]]
df_asia_fil.head(7)
```

```python
X = np.array(df_asia_fil.head(7).values)
X
```

```python
XTX = X.T.dot(X)
```

```python
XTX
```

```python
XTX_inverse = np.linalg.inv(XTX)
XTX_inverse
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = (XTX_inverse.dot(X.T)).dot(y)
w
```

```python
w.sum()
```
